# Premier League 2023/2024

```python
import requests
import json
import pandas as pd

url = 'https://api.football-data.org/v4/competitions/PL/matches/?season=2023'
headers = { 'X-Auth-Token': 'ea822a9b3064455e8e1e8276a4e98ffe' }

response = requests.get(url, headers=headers)

if response.status_code == 200:
    # Convert response to JSON
    data = response.json()

    # print(json.dumps(data, indent=4))
    
    matches = data.get('matches', []) # I dont need to put []. I could just put matches. If the code couldnt find matches, it would throw an error. But by adding [], it will return an empty list ([]) instead of raising an error
    
    if matches:
        # Convert the matches data into a pandas DataFrame
        df = pd.json_normalize(matches)

        # Print DataFrame for inspection (optional)
        print(df)

    
        df.to_excel('PL_Team_Data_Response.xlsx', index=False)

        print("Data saved to football_data_response.xlsx")
    
    else:
        print("No matches found in the response.")
    
else:
    print(f"Error: Received status code {response.status_code}")
```
